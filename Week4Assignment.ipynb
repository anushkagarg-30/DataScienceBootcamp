# Week 4 Assignment - Sales Analysis

```python
import pandas as pd

sales = pd.read_csv("sales.csv", parse_dates=["Date"])
items = pd.read_csv("items.csv")
customers = pd.read_csv("customers.csv")
```

## Q1: Total orders completed on 18th March 2023

```python
orders_18_march = sales[sales['Date'] == '2023-03-18']
total_orders_18_march = orders_18_march['Order_id'].nunique()
print("Total Orders on 18th March 2023:", total_orders_18_march)
```

## Q2: Orders on 18th March 2023 by 'John Doe'

```python
john_doe_ids = customers[(customers['first_name'] == 'John') & (customers['last_name'] == 'Doe')]['Customer_id']
orders_john_doe = sales[(sales['Date'] == '2023-03-18') & (sales['Customer_id'].isin(john_doe_ids))]
total_john_doe_orders = orders_john_doe['Order_id'].nunique()
print("Orders by John Doe on 18th March 2023:", total_john_doe_orders)
```

## Q3: Jan 2023 - Total customers and average spend per customer

```python
jan_2023_sales = sales[sales['Date'].dt.month == 1]
total_customers_jan = jan_2023_sales['Customer_id'].nunique()
avg_spend_jan = jan_2023_sales.groupby('Customer_id')['Revenue'].sum().mean()
print("Total Customers in Jan 2023:", total_customers_jan)
print("Average Spend per Customer:", round(avg_spend_jan, 2))
```

## Q4: Departments with < $600 Revenue in 2022

```python
sales_2022 = sales[sales['Date'].dt.year == 2022]
merged_2022 = sales_2022.merge(items, on='Item_id')
dept_revenue = merged_2022.groupby('department')['Revenue'].sum()
low_depts = dept_revenue[dept_revenue < 600]
print(low_depts)
```

## Q5: Most and Least Revenue by an Order

```python
order_revenue = sales.groupby('Order_id')['Revenue'].sum()
print("Max Revenue by Order:", order_revenue.max())
print("Min Revenue by Order:", order_revenue.min())
```

## Q6: Items in the Most Lucrative Order

```python
top_order_id = order_revenue.idxmax()
top_order_details = sales[sales['Order_id'] == top_order_id].merge(items, on='Item_id')
print(top_order_details[['Order_id', 'Item_name', 'Quantity', 'Revenue']])
```
